```python
import numpy as np
import matplotlib.pyplot as plt
import os

moon_path = os.path.abspath(
    "../Data/Moon_Data/ldem_16_float.img"
)

hgt = np.fromfile(moon_path, dtype=np.float32)

hgt = hgt.reshape(2880, 5760)

print("Shape:", hgt.shape)
print("Data Type:", hgt.dtype)
print("Minimum:", hgt.min())
print("Maximum:", hgt.max())
```

```python
plt.figure(figsize=(16, 8))

plt.imshow(
    hgt,
    cmap='terrain',
    vmin=-9,
    vmax=11
)

plt.colorbar(label='Elevation (km)')

plt.title('Global Lunar Elevation Map')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.hist(
    hgt.flatten(),
    bins=100,
    color='steelblue',
    edgecolor='black',
    alpha=0.7
)

plt.axvline(
    hgt.mean(),
    color='red',
    linestyle='--',
    label=f'Mean: {hgt.mean():.2f} km'
)

plt.axvline(
    np.median(hgt),
    color='orange',
    linestyle='--',
    label=f'Median: {np.median(hgt):.2f} km'
)

plt.title('Distribution of Lunar Elevation')
plt.xlabel('Elevation (km)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```

Elevation Distribution

```python
percentiles = [0, 10, 25, 50, 75, 90, 95, 99, 100]

for p in percentiles:
    print(f"{p}th Percentile:",
          np.percentile(hgt, p))
```

```python
# Interquartile Range
Q1 = np.percentile(hgt, 25)
Q3 = np.percentile(hgt, 75)
IQR = Q3 - Q1

# Additional percentiles
P1 = np.percentile(hgt, 1)
P99 = np.percentile(hgt, 99)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("P1:", P1)
print("P99:", P99)
```
