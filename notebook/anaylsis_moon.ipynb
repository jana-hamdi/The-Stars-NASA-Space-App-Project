```python
img_path = os.path.join(
    project_path,
    "Data",
    "Data_Moon",
    "ldem_16_float.img"
)

moon_data = np.fromfile(img_path, dtype="<f4")

moon_data = moon_data.reshape((2880, 5760))

print("Shape:", moon_data.shape)
print("Data Type:", moon_data.dtype)
print("Total Pixels:", moon_data.size)
```

#### Basic Statistical Analysis

```python
print("Minimum:", np.min(moon_data), "km")
print("Maximum:", np.max(moon_data), "km")
print("Mean:", np.mean(moon_data), "km")
print("Median:", np.median(moon_data), "km")
print("Standard Deviation:", np.std(moon_data), "km")
```
